## Importing your project code

Because the project is a package and `uv sync` installs it **editable**, you
import by the **package name** — never `from src...` and never `sys.path` hacks.

First, make sure Jupyter is running in the project environment:

```bash
uv add --dev jupyterlab ipykernel
uv run jupyter lab        # or select the project's .venv as the kernel in VS Code
```
